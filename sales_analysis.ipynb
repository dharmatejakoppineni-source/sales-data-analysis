```python
###statistical concepts--normalization
##to have a mean of and standaard deviation of 1
data=np.array([1,2,3,4,5])

##caluclate the mean and standard deviation
mean=np.mean(data)
std_dev=np.std(data)

##normalize the data
normalized_data=(data-mean)/std_dev
print("normalized data",normalized_data)

##output:
#normalized data:[-1,26491106-0.63245553 0.0.63245553.1.2641106]
```

```python
query = """
SELECT Country,sum("sum of sales") as total_sales
from financials
group by country
order by total_sales desc
"""
```

```python
result = pd.read_sql(query,conn)
result
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.bar(result['Country'],result['total_sales'],color='blue')
plt.xlabel('Country')
plt.ylabel('Total Sales')
plt.title('country_wise total sales')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(result['total_sales'], labels=result['Country'], autopct='%1.1f%%',startangle=90)
plt.title('Country_wise sales distribution')
plt.axis('equal')
plt.show()
```
